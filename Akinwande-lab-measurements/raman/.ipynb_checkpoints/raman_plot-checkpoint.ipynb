```python
# plots raman data and highlights peaks for graphene
# for graphene MUST have the range between 1000 and 3000

# If using the quantitative measurements, make sure to have baseline subtraction True
```

```python
# dependencies
import matplotlib.pyplot as plt
import os
import numpy as np

from scipy.signal import medfilt
from pybaselines import Baseline
from scipy.signal import peak_widths
from scipy.optimize import curve_fit
```

```python


# DO I NEED A +C TERM FOR THIS TO ACCOUNT FOR Y-MISMATCH
def lorentzian(x, A, w0, Gamma):
    """
    Lorentzian peak function

    A      = peak amplitude
    w0     = center position
    Gamma  = half-width parameter
    """
    return A / (1 + ((x - w0) / Gamma)**2)

def double_lorentzian(x,
                      A1, w01, Gamma1,
                      A2, w02, Gamma2,
                      C):

    L1 = lorentzian(x, A1, w01, Gamma1)
    L2 = lorentzian(x, A2, w02, Gamma2)

    return L1 + L2 + C

    
class RamanProcessor:
    def __init__(
        self,
        data_folder_path,
    ):
        self.data_folder_path = data_folder_path

    def plot_entire_range(self, med_filt_kernel_size=7, baseline_correct=False, normalized=False):

        for filename in os.listdir(os.path.join(data_folder_path)):
            if filename == ".ipynb_checkpoints": continue
            
            print('plotting', filename)

            data = np.loadtxt(os.path.join(os.path.join(data_folder_path), filename))
            x = data[:, 0]
            y = data[:, 1]


            y_filtered = medfilt(y, kernel_size=med_filt_kernel_size)

            if baseline_correct:
                baseline_fitter = Baseline()

                baseline, params = baseline_fitter.asls(
                    # x,
                    y_filtered,
                    lam=1e6,
                    p=0.001
                )

                y_filtered = y_filtered - baseline

            
            normalized_text = " (a.u.)"
            if normalized:
                y_filtered = y_filtered / max(y_filtered)
                normalized_text = " (Normalized)"
            
            peaks_xy = []
            fwhm_vals = []
            graphene_flag = False # for labeling if the raman plot is for graphene or Si
        
            # D peak ~1350
            # G peak ~1580
            # 2D peak ~2700
            for peak_value in [1350, 1580, 2700]:
                try:
                    mask = (x >= peak_value - 100) & (x <= peak_value + 100) # HAD 70 EARLIER
                    x_subset = x[mask]
                    y_subset = y_filtered[mask]
                    max_idx = np.argmax(y_subset)
                    x_at_max_y = x_subset[max_idx]
                    peaks_xy.append((x_at_max_y, np.max(y_subset)))
                    graphene_flag = True

                    # fwhm calculation
                    widths, width_heights, left_ips, right_ips = peak_widths(
                        y_subset,
                        [max_idx],
                        rel_height=0.5
                    )
                    dx = np.mean(np.diff(x_subset)) * -1
                    fwhm = widths[0] * dx
                    fwhm_vals.append(fwhm)
                    
                    
                except:
                    print('ERROR FINDING GRAPHENE PEAKS, TRYING TO DETECT SI PEAK')
                    try:
                        mask = (x >= 520 - 10) & (x <= 520 + 10)
                        x_subset = x[mask]
                        y_subset = y_filtered[mask]
                        max_idx = np.argmax(y_subset)
                        x_at_max_y = x_subset[max_idx]
                        peaks_xy.append((x_at_max_y, np.max(y_subset)))
                    except:
                        print('COULD NOT FIND ANY PEAKS')

            # printing quantitative stats about peaks
            print(peaks_xy)
            
            print('G peak:', peaks_xy[1][0], 'cm-1')
            print('2D peak', peaks_xy[2][0], 'cm-1')
            print('I_2D/I_G', round(peaks_xy[2][1] / peaks_xy[1][1], 1))
            print('I_G/I_D', round(peaks_xy[1][1] / peaks_xy[0][1], 1))
            print('G peak FWHM:', round(fwhm_vals[1], 1))
            print('2D peak FWHM:', round(fwhm_vals[2], 1))

            

            # plot and make pretty
            plt.plot(x, y_filtered)
            plt.xlabel(r'Raman shift (cm $^{-1}$)', fontsize=16)
            plt.ylabel('Intensity' + normalized_text, fontsize=16)
            plt.ylim(0,max(y_filtered)*1.15)
            plt.title(filename)

            
            # if graphene_flag:
            #     peak_names =['D', 'G', '2D']
            #     # plotting peaks
            #     for i, (x_peak, y_peak) in enumerate(peaks_xy):
            #         # plt.scatter(x_peak,x_peak, marker='o', s=50)
            #         plt.plot([x_peak, x_peak], [0, max(y_filtered)*1.15], color='grey', linestyle='--')
            #         plt.text(x_peak, y_peak, f'{peak_names[i]}\n' + str(x_peak), fontsize=10, color='black', ha='center', va='bottom')
            # else:
            #     plt.scatter(peaks_xy[0][0], peaks_xy[0][1], marker='o', s=50)
            #     plt.text(peaks_xy[0][0], peaks_xy[0][1], f'Si\n' + str(peaks_xy[0][0]), fontsize=10, color='black', ha='center', va='bottom')
            plt.show()
            plt.close()

    def plot_G_peak(self, med_filt_kernel_size=7, baseline_correct=False, normalized=False):

        for filename in os.listdir(os.path.join(data_folder_path)):
            if filename == ".ipynb_checkpoints": continue
            data = np.loadtxt(os.path.join(os.path.join(data_folder_path), filename))
            x = data[:, 0]
            y = data[:, 1]

            # mask = (x >= 1400) & (x <= 1800)
            # x = x[mask]
            # y = y[mask]
            # y_filtered = medfilt(y, kernel_size=med_filt_kernel_size)
            # y_filtered = y_filtered / max(y_filtered)

            y_filtered = medfilt(y, kernel_size=med_filt_kernel_size)

            if baseline_correct:
                baseline_fitter = Baseline()

                baseline, params = baseline_fitter.asls(
                    y_filtered,
                    lam=1e6,
                    p=0.001
                )

                y_filtered = y_filtered - baseline

            
            normalized_text = " (a.u.)"
            if normalized:
                y_filtered = y_filtered / max(y_filtered)
                normalized_text = " (Normalized)"

            mask = (x >= 1400) & (x <= 1800)
            x = x[mask]
            y_filtered = y_filtered[mask]


            peaks_xy = []
            graphene_flag = False # for labeling if the raman plot is for graphene or Si
        
            # D peak ~1350
            # G peak ~1580
            # 2D peak ~2700
            for peak_value in [1350, 1580, 2700]:
                try:
                    mask = (x >= peak_value - 70) & (x <= peak_value + 70)
                    x_subset = x[mask]
                    y_subset = y_filtered[mask]
                    max_idx = np.argmax(y_subset)
                    x_at_max_y = x_subset[max_idx]
                    peaks_xy.append((x_at_max_y, np.max(y_subset)))
                    graphene_flag = True
                except:
                    print('ERROR FINDING GRAPHENE PEAKS, TRYING TO DETECT SI PEAK')
                    try:
                        mask = (x >= 520 - 10) & (x <= 520 + 10)
                        x_subset = x[mask]
                        y_subset = y_filtered[mask]
                        max_idx = np.argmax(y_subset)
                        x_at_max_y = x_subset[max_idx]
                        peaks_xy.append((x_at_max_y, np.max(y_subset)))
                    except:
                        print('COULD NOT FIND ANY PEAKS')
        
            # plot and make pretty
            print('plotting', filename)

            # mask = (x >= 1300) & (x <= 1700)
            plt.plot(x, y_filtered)
            plt.xlabel(r'Raman shift (cm $^{-1}$)')
            plt.ylabel('Intensity' + normalized_text)
            plt.ylim(0, max(y_filtered)*1.15)
            plt.title(filename)
            
            plt.show()
            plt.close()

    def plot_G_1_lorenzian(self, med_filt_kernel_size=7, baseline_correct=False, normalized=False):
    
        for filename in os.listdir(os.path.join(data_folder_path)):
            if filename == ".ipynb_checkpoints":
                continue
    
            data = np.loadtxt(os.path.join(data_folder_path, filename))
            x = data[:, 0]
            y = data[:, 1]
    
            # Median filter
            y_filtered = medfilt(y, kernel_size=med_filt_kernel_size)
    
            # Baseline correction
            if baseline_correct:
    
                baseline_fitter = Baseline()
                baseline, params = baseline_fitter.asls(y_filtered, lam=1e6, p=0.001)
                y_filtered = y_filtered - baseline
    
            # Normalize if desired
            normalized_text = " (a.u.)"
            if normalized:
                y_filtered = y_filtered / np.max(y_filtered)
                normalized_text = " (Normalized)"
    
            # ONLY LOOK AT G PEAK REGION
            mask = (x >= 1500) & (x <= 1660)
            x_fit = x[mask]
            y_fit = y_filtered[mask]
    
            # Initial Lorenzian params guess
            if not list(y_fit): continue
            A_guess = np.max(y_fit)
            w0_guess = x_fit[np.argmax(y_fit)]
            Gamma_guess = 15
            p0 = [A_guess, w0_guess, Gamma_guess]
    
            # Fit Lorenzian
            try:
                popt, pcov = curve_fit(
                    lorentzian,
                    x_fit,
                    y_fit,
                    p0=p0
                )
    
                A_fit, w0_fit, Gamma_fit = popt
                y_lorentz = lorentzian(x_fit, *popt)
    
                # Plot
                plt.figure(figsize=(6, 5))
                plt.plot(
                    x_fit,
                    y_fit,
                    label='Experimental Data'
                )

                print(f'\nA = {A_fit:.3f}\n'
                        f'ω₀ = {w0_fit:.2f} cm⁻¹\n'
                        f'Γ = {Gamma_fit:.2f} cm⁻¹')
                
                plt.plot(
                    x_fit,
                    y_lorentz,
                    '--',
                    linewidth=2,
                    label=(
                        f'Lorentzian Fit'
                        # f'\nA = {A_fit:.3f}\n'
                        # f'ω₀ = {w0_fit:.2f} cm⁻¹\n'
                        # f'Γ = {Gamma_fit:.2f} cm⁻¹'
                    )
                )
    
                plt.xlabel(r'Raman shift (cm$^{-1}$)', fontsize=16)
                plt.ylabel('Intensity' + normalized_text, fontsize=16)
                plt.title(f'G Peak Fit: {filename}')
                plt.legend()
                plt.tight_layout()
                plt.show()
                plt.close()
            except RuntimeError:
                print(f'Fit failed for {filename}')




    def plot_G_2_lorenzian(self,
                               med_filt_kernel_size=7,
                               baseline_correct=False,
                               normalized=False):
    
        for filename in os.listdir(data_folder_path):
    
            if filename == ".ipynb_checkpoints":
                continue
    
            print(f"\nProcessing: {filename}")
    
            # ----------------------------------------
            # LOAD DATA
            # ----------------------------------------
            filepath = os.path.join(data_folder_path, filename)
    
            data = np.loadtxt(filepath)
    
            x = data[:, 0]
            y = data[:, 1]
    
            # ----------------------------------------
            # FILTER
            # ----------------------------------------
            y_filtered = medfilt(y, kernel_size=med_filt_kernel_size)
    
            # ----------------------------------------
            # BASELINE CORRECTION
            # ----------------------------------------
            if baseline_correct:
    
                baseline_fitter = Baseline()
    
                baseline, params = baseline_fitter.asls(
                    y_filtered,
                    lam=1e6,
                    p=0.001
                )
    
                y_filtered = y_filtered - baseline
    
            # ----------------------------------------
            # NORMALIZE
            # ----------------------------------------
            ylabel_text = "Intensity (a.u.)"
    
            if normalized:
                y_filtered = y_filtered / np.max(y_filtered)
                ylabel_text = "Intensity (Normalized)"
    
            # ----------------------------------------
            # G PEAK WINDOW
            # ----------------------------------------
            mask = (x >= 1500) & (x <= 1660)

    
            x_fit = x[mask]
            y_fit = y_filtered[mask]

            if not list(x_fit): 
                print('Skipping ')
                continue
    
            # ----------------------------------------
            # INITIAL GUESSES
            # ----------------------------------------
            peak_idx = np.argmax(y_fit)
    
            peak_center = x_fit[peak_idx]
            peak_height = y_fit[peak_idx]
    
            # Guess two nearby peaks
            p0 = [
                peak_height * 0.7, peak_center - 5, 8,   # Peak 1
                peak_height * 0.3, peak_center + 5, 12,  # Peak 2
                np.min(y_fit)                            # Constant offset
            ]
    
            # ----------------------------------------
            # BOUNDS
            # ----------------------------------------
            lower_bounds = [
                0, 1540, 1,
                0, 1540, 1,
                -np.inf
            ]
    
            upper_bounds = [
                np.inf, 1620, 50,
                np.inf, 1620, 50,
                np.inf
            ]
    
            # ----------------------------------------
            # FIT
            # ----------------------------------------
            try:
    
                popt, pcov = curve_fit(
                    double_lorentzian,
                    x_fit,
                    y_fit,
                    p0=p0,
                    bounds=(lower_bounds, upper_bounds),
                    maxfev=20000
                )
    
                # Unpack parameters
                A1, w01, G1, A2, w02, G2, C = popt
    
                # Individual components
                L1 = lorentzian(x_fit, A1, w01, G1)
                L2 = lorentzian(x_fit, A2, w02, G2)
    
                # Total fit
                y_total = double_lorentzian(x_fit, *popt)
    
                # ----------------------------------------
                # PRINT PARAMETERS
                # ----------------------------------------
                print("\nPeak 1")
                print(f"A1     = {A1:.3f}")
                print(f"w01    = {w01:.3f} cm^-1")
                print(f"Gamma1 = {G1:.3f} cm^-1")
    
                print("\nPeak 2")
                print(f"A2     = {A2:.3f}")
                print(f"w02    = {w02:.3f} cm^-1")
                print(f"Gamma2 = {G2:.3f} cm^-1")
    
                # ----------------------------------------
                # PLOT
                # ----------------------------------------
                plt.figure(figsize=(6, 5))
    
                # Experimental data
                plt.plot(
                    x_fit,
                    y_fit,
                    linewidth=2,
                    label='Experimental Data'
                )
    
                # Total fit
                plt.plot(
                    x_fit,
                    y_total,
                    '--',
                    linewidth=3,
                    label='Double-Lorentzian Fit'
                )
    
                # Individual Lorentzians
                plt.plot(
                    x_fit,
                    L1 + C,
                    ':',
                    linewidth=2,
                    label=(
                        f'G Peak Lorenzian'
                        # f'\nω₀ = {w01:.2f}\n'
                        # f'Γ = {G1:.2f}'
                    )
                )
    
                plt.plot(
                    x_fit,
                    L2 + C,
                    ':',
                    linewidth=2,
                    label=(
                        f'D` Peak Lorenzian'
                        # f'\nω₀ = {w02:.2f}\n'
                        # f'Γ = {G2:.2f}'
                    )
                )
    
                plt.xlabel(r'Raman Shift (cm$^{-1}$)', fontsize=16)
                plt.ylabel(ylabel_text, fontsize=16)
    
                plt.title(f'Double Lorentzian G Peak Fit: {filename}')
                plt.legend() #fontsize=16)
                plt.tight_layout()
                plt.show()
                plt.close()
            except Exception as e:
                print(f"Fit failed for {filename}")
                print(e)

            
```
